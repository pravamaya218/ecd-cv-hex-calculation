```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

rectangles = [
    (704, 528, 518, 389),
    (72, 528, 518, 389),
    (704, 56, 518, 389),
    (72, 56, 518, 389)
]

fig, ax = plt.subplots(figsize=(10, 10))

for (x, y, w, h) in rectangles:
    rect = patches.Rectangle((x, y), w, h, linewidth=2, edgecolor='blue', facecolor='none')
    ax.add_patch(rect)
    ax.text(x, y, f"({x},{y})", fontsize=10, color='red')

ax.set_xlim(0, 1300)
ax.set_ylim(0, 1000)
ax.set_aspect('equal')
plt.gca().invert_yaxis()  # optional: matches typical image coordinate systems
plt.show()
```
